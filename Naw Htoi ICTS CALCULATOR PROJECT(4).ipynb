## Project Description

This project is to build a basic calculator that can perform addition, subtraction, multiplication, division, second power, and square root.
The user can pick the operations from a menu and repeat this action until they wish to stop.
Each operation is implemented as its own function for convenience and simplicity.

### First Part

In the code cell below, I will be creating basic operations such as addition, subtraction, multiplication, and division.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
```

In the next code cell, I will be implementing the harder operations, square root and second power.

```python
def power(a, b):
    return a ** b

def square_root(n):
    if n < 0:
        raise ValueError("Cannot take square root of a negative number")
    return n ** 0.5
```

### Second Part

Within the following cells, I will create the menu, which will be a very simple menu that asks the user what operation they would like to use, before asking the two numbers and then showing the result. I will also add in the looping function and the exit function at the end.

```python
def show_menu():
    """Calculator Menu"""
    print("\n Calculator ")
    print("1. Addition ")
    print("2. Subtraction ")
    print("3. Multiplication ")
    print("4. Division ")
    print("5. Second power ")
    print("6. Square root ")
    print("7. Exit ")
```

```python
def main():
    """Run calculator until user chooses exit."""
    while True:
        show_menu()
        choice = input("Choose an option: ")

        if choice == '7':
            print("Exited")
            break

        try:
            if choice == '6':
                n = float(input("Enter a number: "))
                print(f"Square root of {n} = {square_root(n):.2f}")
                
            elif choice in ('1', '2', '3', '4', '5'):
                a = float(input("First number:  "))
                b = float(input("Second number: "))
                operations = { '1': add, '2': subtract, '3': multiply, '4': divide, '5': power }
                print(f"Result: {operations[choice](a, b)}")
                
            else:
                print("Invalid choice. Enter 1-7.")
                
        except (ValueError, ZeroDivisionError) as e:
            print(f"Error: {e}")
```
